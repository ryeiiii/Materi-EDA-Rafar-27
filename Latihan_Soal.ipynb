```python
import pandas as pd
import numpy as np

# ==========================================
# TAHAP 1: DATA LOADING & INSPECTION
# ==========================================
print("--- 1. DATA LOADING & INSPECTION ---")
df = pd.read_csv('dataset_penjualan_kantin.csv')

print("Info Dataset Awal:")
df.info()

print("\nJumlah Data Kosong (Missing Values):")
print(df.isnull().sum())

print("\nJumlah Data Duplikat:", df.duplicated().sum())
print("-" * 50)


# ==========================================
# TAHAP 2: DATA CLEANING
# ==========================================
print("\n--- 2. DATA CLEANING ---")

# 2.1 Menghapus baris duplikat
df = df.drop_duplicates()

# 2.2 Membersihkan kolom 'jumlah_terjual' (menghapus kata "pcs")
df['jumlah_terjual'] = df['jumlah_terjual'].astype(str).str.replace(' pcs', '').str.strip()
df['jumlah_terjual'] = df['jumlah_terjual'].replace('nan', np.nan).astype(float)
# Mengisi jumlah_terjual yang kosong dengan angka 0 lalu ubah ke integer
df['jumlah_terjual'] = df['jumlah_terjual'].fillna(0).astype(int)

# 2.3 Membersihkan kolom 'harga_satuan' (menghapus "Rp" dan ".")
df['harga_satuan'] = df['harga_satuan'].astype(str).str.replace('Rp', '', regex=False).str.replace('.', '', regex=False).str.strip()
df['harga_satuan'] = df['harga_satuan'].replace('nan', np.nan).astype(float)

# 2.4 Menangani Missing Value sisa (menghapus baris jika harga atau kasir kosong)
df = df.dropna(subset=['harga_satuan', 'nama_kasir'])
df['harga_satuan'] = df['harga_satuan'].astype(int)

# 2.5 Menyeragamkan penulisan pada kolom kategori (Makanan, Minuman, Snack)
df['kategori'] = df['kategori'].str.title()

print("\nInfo Dataset Setelah Dibersihkan:")
df.info()
print("-" * 50)


# ==========================================
# TAHAP 3: DATA MANIPULATION
# ==========================================
print("\n--- 3. DATA MANIPULATION ---")

# 3.1 Membuat kolom turunan
df['total_pendapatan'] = df['jumlah_terjual'] * df['harga_satuan']

# 3.2 Filtering (Mengambil data yang produknya benar-benar terjual > 0)
df_valid = df[df['jumlah_terjual'] > 0]

# 3.3 Menjawab Pertanyaan 1: Menu Pendapatan Tertinggi (Groupby & Sorting)
menu_tertinggi = df_valid.groupby('nama_produk')['total_pendapatan'].sum().reset_index()
menu_tertinggi = menu_tertinggi.sort_values(by='total_pendapatan', ascending=False)
print("\nMenu dengan Total Pendapatan Tertinggi:")
print(menu_tertinggi.head())

# 3.4 Menjawab Pertanyaan 2: Kategori Paling Laris
kategori_terlaris = df_valid.groupby('kategori')['jumlah_terjual'].sum().reset_index()
kategori_terlaris = kategori_terlaris.sort_values(by='jumlah_terjual', ascending=False)
print("\nKategori Paling Banyak Terjual (Kuantitas):")
print(kategori_terlaris)

# 3.5 Menjawab Pertanyaan 3: Kasir dengan Pendapatan Terbanyak
kasir_terbaik = df_valid.groupby('nama_kasir')['total_pendapatan'].sum().reset_index()
kasir_terbaik = kasir_terbaik.sort_values(by='total_pendapatan', ascending=False)
print("\nKasir dengan Catatan Pendapatan Terbanyak:")
print(kasir_terbaik.head())

# Opsional: Menyimpan data bersih untuk visualisasi (Elemen 3)
# df_valid.to_csv('dataset_penjualan_kantin_bersih.csv', index=False)
```
